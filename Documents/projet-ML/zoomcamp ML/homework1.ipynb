```python
X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].iloc[:7].to_numpy()

print(X)
```

```python
XTX = X.T @ X
print(X) #before Transpose
print(X.T) #Transpose
print(XTX) #result
```

```python
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
print(y)
```

```python
w = XTX_inv @ X.T @ y
print(w)
```
